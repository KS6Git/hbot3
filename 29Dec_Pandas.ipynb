```python
#df['Marks'].apply(lambda x:'Pass' if x>=50 else 'Fail')
def new_fun(x):
  if x>=50:
    return 'Pass'
  else:
    return 'Fail'
```

```python
# same as writing the prog above in a single line
df['Marks'].apply(lambda x:'Pass' if x>=50 else 'Fail')
```

```python
# the entire prog is def as new_fun does the same thing
df['Marks'].apply(new_fun)
```

```python
# appending
df['Result'] = df['Marks'].apply(new_fun)
```
